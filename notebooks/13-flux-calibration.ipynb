# Flux Calibration

:::{warning}
**This chapter is a placeholder.** `specreduce` gains its flux-calibration tools in **v1.11**, and
this chapter will be written in full against them.
:::

Chapter 12 left the reduction with a spectrum on a physical wavelength axis and an instrumental
vertical one. The flux is in electrons per Ångström: detected charge, not energy from the sky.
Everything between the source and the detector, the atmosphere, the telescope optics, the slit,
the grism and the detector's quantum efficiency, has a wavelength-dependent throughput, and their
product, the **instrument response**, is stamped on every spectrum the book has produced.

Flux calibration divides that response out and converts the spectrum into physical units,
conventionally erg s⁻¹ cm⁻² Å⁻¹. It needs one more observation: a **spectrophotometric standard
star**, whose true flux distribution is tabulated. Reduce the standard exactly as the target was
reduced, divide its observed spectrum by its tabulated one, and the ratio is the **sensitivity
function**, the conversion between counts and physical flux at each wavelength. Applied to the
target, it removes the response.
